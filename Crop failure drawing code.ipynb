# Line chart

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.ticker import MaxNLocator
import seaborn as sns

sns.set(font='Times New Roman')

merged_data = pd.read_excel("")

merged_data.replace(0, np.nan, inplace=True)

colors = ['#DE0F0F','#61C6E7','#F6A400','#8827A4','#F6A400','#DE0F0F','#7ABF57','#F6A400','#DE0F0F','#F6A400']

i = 0
for col in merged_data.columns:
    if 'failure' in col:
        
        anomaly_col = col.replace('-failure', '-anomaly')
        
        if anomaly_col in merged_data.columns:
            plt.figure(figsize=(8, 4))
            plt.plot(merged_data['year_x'], merged_data[col]/-10000, linestyle='-', linewidth=1.5, marker='o', markersize=5, label=f'{col}',color=colors[i])
            plt.plot(merged_data['year_x'], merged_data[anomaly_col]/10000, linestyle='-', linewidth=3, alpha=0.2, label=f'{anomaly_col} ',color=colors[i])

            legend = plt.legend(fontsize='16')
            legend.get_frame().set_edgecolor('none')
            legend.get_frame().set_facecolor('none')
            
            plt.gca().set_facecolor('none')  
            plt.gca().spines['top'].set_color('black')
            plt.gca().spines['right'].set_color('black')
            plt.gca().spines['bottom'].set_color('black')
            plt.gca().spines['left'].set_color('black')
            plt.ylim(top=0)
            plt.xlim(left=1980, right=2019)
            plt.gca().yaxis.set_major_locator(MaxNLocator(integer=True))
            plt.tick_params(bottom=True, top=False, left=True, right=False,axis='both', which='both', direction='in', labelsize=20)  # Adjust tick parameters
            plt.grid(False)
            last_sheet_name = col
            sheet_name_parts = last_sheet_name.split('-')  
            if len(sheet_name_parts) > 1:
                new_sheet_name = '-'.join(sheet_name_parts[:-1])  
            else:
                new_sheet_name = last_sheet_name  
            plt.savefig("F:\\" + new_sheet_name + ".png", bbox_inches='tight',dpi = 600)

            plt.show()
            i = i + 1
```

# 3D Visualization

```python
import numpy as np 
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

sns.set(style="white")
# sns.set(font='Times New Roman')
block_size = [1, 1, 1]  

blocks = [
    {'position': [5, 5, 4], 'size': block_size},
    {'position': [5, 5, 5], 'size': block_size},
    {'position': [5, 4, 4], 'size': block_size},
    {'position': [6, 5, 6], 'size': block_size},
    {'position': [4, 5, 5], 'size': block_size},
]

fig = plt.figure(figsize=(10, 10))
ax = fig.add_subplot(111, projection='3d')

for block in blocks:
    position = block['position']
    size = block['size']
    ax.bar3d(position[0], position[1], position[2], size[0], size[1], size[2], shade=True, color='gray') 

for block in blocks:
    position = block['position']
    size = block['size']
    ax.bar3d(position[0], position[1], 0, size[0], size[1], 0, shade=False, color='r', alpha=0.3)
    ax.bar3d(position[0], 10, position[2], size[0], 0, size[2], shade=False, color='g', alpha=0.3)
    ax.bar3d(0, position[1], position[2], 0, size[1], size[2], shade=False, color='b', alpha=0.3)

ax.set_xlabel('Lat', fontsize=18)
ax.set_ylabel('Lon', fontsize=18)
ax.set_zlabel('Time', fontsize=18)

ax.set_xlim([0, 10])
ax.set_ylim([0, 10])
ax.set_zlim([0, 10])

ax.set_facecolor('none')

ax.set_frame_on(True)  

ax.xaxis.pane.fill = False  
ax.yaxis.pane.fill = False
ax.zaxis.pane.fill = False

ax.grid(True)#False

ax.tick_params(axis='both', which='both', direction='in', labelsize=16)  # 设置刻度
plt.tight_layout()
plt.savefig(r'F:\3d_blocks.png', dpi=600, bbox_inches='tight')
plt.show()
```
